```python
# First simple LLM workflow, where we provide our question to llm and llm will return answer
```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint 
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct"  # model name
)

model = ChatHuggingFace(llm=llm)  # Create a LangChain chat model using the Hugging Face endpoint.
```

```python
# create state class
class LLMState(TypedDict):
    question : str
    answer : str
```

```python
def call_to_llm(state:LLMState)->LLMState:
    #fetch user question
    question = state['question']
    #convert it to prompt
    prompt = f"Answer this question : {question}"
    #give input to model
    answer = model.invoke(prompt).content
    #update state
    state['answer'] = answer
    return state
```

```python
graph = StateGraph(LLMState)
graph.add_node('call_to_llm',call_to_llm)
graph.add_edge(START,'call_to_llm')
graph.add_edge('call_to_llm',END)

workflow = graph.compile()
```

```python
initial_state = {'question': 'who is virat kohli'}
final_state = workflow.invoke(initial_state)
print(final_state)
```
